```python

# Clock Master Project
# Step 3: Digital Clock

# Import datetime to access the current date and time
from datetime import datetime #Imports a Python class for working with dates and times.

# Import time to pause the program
import time #Imports time-related functions.


# Define a function for the digital clock
def digital_clock():

    # Display the module heading
    print("\n--- DIGITAL CLOCK ---")

    # Keep displaying the time until the user stops the cell
    try:
        while True:

            # Get the current date and time from the system
            current_time = datetime.now()

            # Display the date and time in a readable format =The format "%d-%m-%Y %H:%M:%S" means day-month-year followed by hour-minute-second.
            print(current_time.strftime("%d-%m-%Y %H:%M:%S"))

            # Wait for one second before displaying again,Pauses the program for one second.
            time.sleep(1)

    # Handle the user interrupting the running cell
    except KeyboardInterrupt:
        print("Digital clock stopped.")
```

```python
# Calling the digital clock function
digital_clock()
```

```python
# Step 4: Alarm Clock
# Define the alarm clock function
def alarm_clock(alarm_time):
    print(f"\n--- ALARM SET FOR {alarm_time} ---")
    print("Waiting for the alarm... (Press Ctrl+C to cancel)")

    try:
        while True:
            current_time = datetime.now().strftime("%H:%M:%S")

            print(f"Current Time: {current_time}", end="\r")

            if current_time == alarm_time:
                print("\n\n⏰ WAKE UP! ALARM RINGING! ⏰")
                print("\a")
                break

            time.sleep(1)

    except KeyboardInterrupt:
        print("\nAlarm cancelled.")
```

```python

# Call the alarm clock function
alarm_clock("10:25:00")
```

```python
# Step 5: Stopwatch

# Import the time module to measure elapsed time
def stopwatch():
    print("\n--- STOPWATCH STARTED ---")
    print("(Press Ctrl+C to stop)")

    seconds_elapsed = 0

    try:
        while True:
            m, s = divmod(seconds_elapsed, 60)
            h, m = divmod(m, 60)

            current_time = f"{h:02d}:{m:02d}:{s:02d}"
            print(f"Elapsed Time: {current_time}", end="\r")

            time.sleep(1)
            seconds_elapsed += 1

    except KeyboardInterrupt:
        print("\nStopwatch stopped.")
```

```python
# Call the stopwatch function
stopwatch()
```

```python
# Function to create a countdown timer
def timer(total_seconds):
    print("\n--- TIMER STARTED ---")
    print("(Press Ctrl+C to cancel)")

    try:
        while total_seconds >= 0:
            m, s = divmod(total_seconds, 60)
            h, m = divmod(m, 60)

            time_left = f"{h:02d}:{m:02d}:{s:02d}"
            print(f"Time Remaining: {time_left}", end="\r")

            if total_seconds == 0:
                print("\n\n⏰ TIME'S UP! ⏰\a")
                break

            time.sleep(1)
            total_seconds -= 1

    except KeyboardInterrupt:
        print("\nTimer cancelled.")
```

```python
timer(60)
```

```python

# Main function to connect all Clock Master features
def main():
    while True:
        print("\n================================")
        print("         CLOCK MASTER")
        print("================================")
        print("1. Digital Clock")
        print("2. Alarm Clock")
        print("3. Stopwatch")
        print("4. Countdown Timer")
        print("5. Exit")

        # Get the user's menu choice
        choice = input("Enter your choice (1-5): ")

        # Call the selected function
        if choice == "1":
            digital_clock()
        elif choice == "2":
            alarm_time = input("Enter alarm time (HH:MM:SS): ")
            alarm_clock(alarm_time)
        elif choice == "3":
            stopwatch()
        elif choice == "4":
            user_seconds = int(input("Enter time in seconds: "))
            timer(user_seconds)
        elif choice == "5":
            print("Thank you for using Clock Master!")
            break
        else:
            print("Invalid choice! Please enter a number from 1 to 5.")
```

```python
main()
```
